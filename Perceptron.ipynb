```python
import numpy as np

class PerceptronNumpy:
    def __init__(self, input_size, lr=0.1, epochs=1000):
        # poids aléatoires + biais
        self.weights = np.random.randn(input_size)
        self.bias = np.random.randn()
        self.lr = lr
        self.epochs = epochs

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def predict(self, X):
        # X dot W + b en une ligne
        z = np.dot(X, self.weights) + self.bias
        return self.sigmoid(z)

    def train(self, X, y):
     for epoch in range(self.epochs):
        # 1. Forward
        output = self.predict(X)

        # 2. Erreur + Log Loss
        loss = -np.mean(y*np.log(output + 1e-9) + (1-y)*np.log(1-output + 1e-9))

        # 3. Gradient CORRECT
        gradient = output - y # C'est ça la bonne formule avec LogLoss

        # 4. Update
        self.weights -= self.lr * np.dot(X.T, gradient) # Note le -=
        self.bias -= self.lr * np.sum(gradient)
        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {loss:.4f}") #Affiche la loss avec 4 chiffres après la virgule


X = np.array([[0,0], [0,1], [1,0], [1,1]]) # 4 exemples, 2 features y = np.array([0, 1, 1, 0]) # XOR
y = np.array([1, 0, 1, 0])
p = PerceptronNumpy(input_size=2, lr=0.5, epochs=10000)
p.train(X, y)


for i in range(len(X)):
        print(f"{X[i]} -> {p.predict(X[i]):.3f}")
```

```python
import numpy as np

class Perceptron:
    def __init__(self, learning_rate=0.01, n_iters=1000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.w = None
        self.b = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.n_iters):
            for idx, x_i in enumerate(X):
                linear_output = np.dot(x_i, self.w) + self.b
                y_pred = np.where(linear_output >= 0, 1, -1)
                update = self.lr * (y[idx] - y_pred)
                self.w += update * x_i
                self.b += update

    def predict(self, X):
        linear_output = np.dot(X, self.w) + self.b
        return np.where(linear_output >= 0, 1, -1)

# Exemple
X = np.array([[3,4], [1,4], [2,3], [6,-1], [7,-1]])
y = np.array([1, 1, 1, -1, -1])
perceptron = Perceptron()
perceptron.fit(X, y)
print(perceptron.predict([[2,2], [5,5]]))  # [1 -1]
```
