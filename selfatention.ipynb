```python
import numpy as np  # Importe la bibliothèque numpy pour les calculs matriciels

class SelfAttention:  # Définit une classe pour un mécanisme d'attention
    def __init__(self, input_dim, d_model):
        # Initialise les poids avec des valeurs aléatoires
        self.WQ = np.random.randn(input_dim, d_model)  # Poids Query (entrée → Q)
        self.WK = np.random.randn(input_dim, d_model)  # Poids Key (entrée → K)
        self.WV = np.random.randn(input_dim, d_model)  # Poids Value (entrée → V)
        self.d_model = d_model  # Dimension des embeddings (Q, K, V)

    def scaled_dot_product_attention(self, Q, K, V):
        # Calcul de l'attention (Scaled Dot-Product)
        scores = Q @ K.T / np.sqrt(self.d_model)  # Produit scalaire + scaling (seq_len, seq_len)
        weights = np.softmax(scores, axis=1)       # Poids d'attention (seq_len, seq_len)
        return weights @ V                         # Somme pondérée des V (seq_len, d_model)

    def forward(self, x):
        # x : (seq_len, input_dim) → calcul de Q, K, V
        Q = x @ self.WQ  # (seq_len, d_model)
        K = x @ self.WK  # (seq_len, d_model)
        V = x @ self.WV  # (seq_len, d_model)
        return self.scaled_dot_product_attention(Q, K, V)  # Applique l'attention

    def predict(self, sequence):
        # Prédiction sur une séquence d'entrées
        outputs = []  # Liste pour stocker les sorties
        for x in sequence:
            x_onehot = np.zeros((1, len(self.WQ)))  # Vecteur one-hot (1, input_dim)
            x_onehot[0, x] = 1  # Met à 1 l'index correspondant à x
            output = self.forward(x_onehot)  # (1, d_model)
            outputs.append(output)
        return outputs

# Exemple
vocab_size = 5  # Taille du vocabulaire
d_model = 2     # Dimension des embeddings
attention = SelfAttention(vocab_size, d_model)  # Crée un mécanisme d'attention

# Test
sequence = [1, 2]  # Séquence d'entrée ("hello")
outputs = attention.predict(sequence)
print(outputs)  # Affiche les sorties
```
